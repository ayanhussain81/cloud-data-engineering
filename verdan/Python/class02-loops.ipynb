# Loops


```python
# For Loop

# Syntax:

# for variable in iterable:
#     # Code block to be executed for each item in the iterable

engineers = ["Ayan Hussain", "Qasim Hassan", "Danial", "Moiz"]

for engineer in engineers:
    print(f"Sending report to: {engineer}")
```

```python
# range() function -- return values
# range(10) --> 0 to 9


for i in range(10):
    print (i)

```

```python
tools = ['AWS', 'GCP', 'Azure', 'Terraform', 'Ansible', 'Airflow', 'Docker', 'Kubernetes']
for tool in tools:
    print(tool)
```

```python
for i in range(10, 15):
    print(i)
```

```python
tool = 'Snowflake'
for char in tool:
    print(char)
```

```python
tools = ['AWS', 'Azure', 'PySpark', 'Airflow', 'Docker', 'Kubernetes', 'Snowflake', 'Databricks']

for i in range(len(tools)):
    print(f"{i+1}: {tools[i]}")
```

```python
# Looping through a dictionary

engineer = {
    "name": "Ayan Hussain",
    "role": "Data Engineer",
    "certifications": 4,
    "location": "Karachi"
}

# Loop through keys
for key in engineer.keys():
    print(key)
```

```python
for value in engineer.values():
    print(value)
```

```python
for key, value in engineer.items():
    print(f"{key}: -> {value}")
```
