# Loops


```python
# For Loop

# Syntax:

# for variable in iterable:
#     # Code block to be executed for each item in the iterable

engineers = ["Ayan Hussain", "Qasim Hassan", "Danial", "Moiz"]

for engineer in engineers:
    print(f"Sending report to: {engineer}")
```

```python
# range() function -- return values
# range(10) --> 0 to 9


for i in range(10):
    print (i)

```

```python
tools = ['AWS', 'GCP', 'Azure', 'Terraform', 'Ansible', 'Airflow', 'Docker', 'Kubernetes']
for tool in tools:
    print(tool)
```

```python
for i in range(10, 15):
    print(i)
```

```python
tool = 'Snowflake'
for char in tool:
    print(char)
```

```python
tools = ['AWS', 'Azure', 'PySpark', 'Airflow', 'Docker', 'Kubernetes', 'Snowflake', 'Databricks']

for i in range(len(tools)):
    print(f"{i+1}: {tools[i]}")
```

```python
# Looping through a dictionary

engineer = {
    "name": "Ayan Hussain",
    "role": "Data Engineer",
    "certifications": 4,
    "location": "Karachi"
}

# Loop through keys
for key in engineer.keys():
    print(key)
```

```python
for value in engineer.values():
    print(value)
```

```python
for key, value in engineer.items():
    print(f"{key}: -> {value}")
```

```python
matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42],
]

num = 0
for row in matrix:
    for col in row:
        num += col
    print(f'Sum of row {matrix.index(row) + 1} = {num}')
    num = 0

        
```

```python
sentence = input()
sentence = sentence.lower()
print(sentence)

sentence = sentence.split(" ")
print(sentence)

output = {}

for word in sentence:
    if word in output:
        output[word] += 1
    else:
        output[word] = 1

print(output)
```

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(j * i, end=' ')
    print()
```

```python
word = input('Enter any word to check if its palindrome or not')

palindrome = ''
word_size = len(word) 
for i in range(word_size):
    palindrome += word[word_size-1 - i]

print(palindrome)
if palindrome == word:
    print(f'{word} is palindrome')
else:
    print(f'{word} is not a palindrome')
```
