# Conditional Statements

```python
# Syntac

# if condition:
#     code
# elif:
#     code
# else :
#     code
```

```python
tools = ['AWS', 'Azure', 'PySpark', 'Airflow', 'Docker', 'Kubernetes', 'Snowflake', 'Databricks']

if 'GCP' in tools:
    print("AWS is in the list of tools")
    
elif 'Snowflake' in tools:
    print("Snowflake is in the list of tools")
    
else:
    print("None of the tools are in the list")
```

```python
if 'AWS' in tools and 'Snowflake' in tools:
    print("Both AWS and Snowflake are in the list of tools")
```

# Break


```python
for i in range(10):
    print(i)
    if i == 5:
        break
```

# Contine


```python
for i in range(10):
    if i == 5 or i == 7:
        continue
    print(i)
```

```python
while True:
    print("This is an infinite loop")

## infinite loop
```

```python
num = 0;

while num < 5:
    num += 1
    print(num)

    if num == 3:
        continue
```

```python
retries = 0
max_retries = 3
connected = False

while not connected and retries < max_retries:
    print(f"Attempting connection... (attempt {retries + 1})")
    retries += 1
    if retries == 2:
        connected = True

if connected:
    print("Connection established.")
else:
    print("Connection failed after maximum retries.")
```

# Nested Loops

```python
for tool in tools:
    print(tool)
    for j in tool:
        print(j, end=' ')
    print()
    
```

```python
for i in range(5):
    for j in range(5):
        print('*', end='')
    print()
```

```python
for i in range(5):
    for j in range(i+1):
        print('*', end=' ')
    print()
```

```python
for i in range(5):
    for j in range(5-i+1):
        print(' ', end=' ')
    for j in range(i+1):
        print('*', end=' ')
    print(end='\n')    
    
```
