# Object Oriented Programming

```python
class BankAccount:
    """A simple bank account."""

    # __init__ # constructor / 

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds."
        self.balance -= amount
        return f"Withdrawn: {amount}"

    def summary(self):
        return f"{self.owner}: ${self.balance:,.2f}"
```

```python
vardan = BankAccount('Vardan', 10000000) # class object
akash = BankAccount('Akash', 2000000)

print(vardan.balance)
```

```python
vardan.withdraw(19999)
print(vardan.balance)
print(vardan.summary())
```

```python
# OOP : Inheritence


class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def speak(self):
        return f"{self.name} makes a sound."


class Dog(Animal):
    def speak(self):                    # overrides the parent method
        return f"{self.name} says: Woof!"


dog = Dog("Bruno")
print(dog.eat())      # inherited from Animal
print(dog.speak())    # overridden in Dog
```

```python
class Vehicle:
    def __init__(self, name, brand, model):
        self.name = name
        self.brand = brand
        self.model = model

    def details(self):
        return f'{self.name} | {self.brand} | {self.model}'
    

class Car(Vehicle):
    def __init__(self, name, brand, model, doors):
        super().__init__(name, brand, model)
        self.doors = doors

class Truck(Vehicle):
    def __init__(self, name, brand, model, tyres):
        super().__init__(name, brand, model)
        self.tyres = tyres

Lexus = Car('Lexus 570h', "Lexus", 2026, 2)
print(Lexus.details())
```
